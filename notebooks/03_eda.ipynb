# 03 – Exploratory Data Analysis (EDA)

Visualise engagement patterns across content types, topics, days, and hours.


```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ml.preprocessing import load_and_clean

sns.set_theme(style='whitegrid')
df = load_and_clean('../data/sample/sample_posts.csv')
print(df.shape)
```

```python
# Engagement rate distribution
plt.figure(figsize=(10, 4))
sns.histplot(df['engagement_rate'], bins=40, kde=True, color='steelblue')
plt.title('Distribution of Engagement Rate')
plt.xlabel('Engagement Rate (%)')
plt.tight_layout()
plt.show()
```

```python
# Engagement by content type
plt.figure(figsize=(10, 5))
order = df.groupby('content_type')['engagement_rate'].mean().sort_values(ascending=False).index
sns.boxplot(data=df, x='content_type', y='engagement_rate', order=order, palette='Set2')
plt.title('Engagement Rate by Content Type')
plt.ylabel('Engagement Rate (%)')
plt.tight_layout()
plt.show()
```

```python
# Engagement by topic
plt.figure(figsize=(12, 5))
order = df.groupby('topic')['engagement_rate'].mean().sort_values(ascending=False).index
sns.barplot(data=df, x='topic', y='engagement_rate', order=order, palette='muted', estimator='mean')
plt.title('Average Engagement Rate by Topic')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Engagement by day of week
day_order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='day_of_week', y='engagement_rate', order=day_order, palette='Blues_d')
plt.title('Average Engagement Rate by Day of Week')
plt.tight_layout()
plt.show()
```

```python
# Engagement by posting hour
hour_avg = df.dropna(subset=['posting_hour']).groupby('posting_hour')['engagement_rate'].mean()
plt.figure(figsize=(12, 4))
hour_avg.plot(kind='bar', color='coral')
plt.title('Average Engagement Rate by Posting Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Avg Engagement Rate (%)')
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap (numeric features)
num_cols = ['caption_length','hashtag_count','followers','engagement_rate']
plt.figure(figsize=(7, 5))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
