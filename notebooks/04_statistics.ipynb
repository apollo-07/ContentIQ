# 04 – Statistical Analysis

Descriptive statistics, percentile thresholds, and group comparisons.


```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import numpy as np
from scipy import stats
from ml.preprocessing import load_and_clean

df = load_and_clean('../data/sample/sample_posts.csv')
df.describe()
```

```python
# Performance label distribution
print('Performance label counts:')
print(df['performance_label'].value_counts())
print()

# Percentile thresholds
p33 = df['engagement_rate'].quantile(0.33)
p67 = df['engagement_rate'].quantile(0.67)
print(f'LOW  threshold  (<= {p33:.2f}%)')
print(f'HIGH threshold  (>  {p67:.2f}%)')
```

```python
# Mean engagement per content type with confidence intervals
summary = df.groupby('content_type')['engagement_rate'].agg(['mean','std','count'])
summary['ci_95'] = 1.96 * summary['std'] / np.sqrt(summary['count'])
summary.sort_values('mean', ascending=False)
```

```python
# One-way ANOVA: is there a significant difference between content types?
groups = [g['engagement_rate'].dropna().values for _, g in df.groupby('content_type')]
f_stat, p_val = stats.f_oneway(*groups)
print(f'ANOVA F-statistic: {f_stat:.2f}, p-value: {p_val:.4f}')
if p_val < 0.05:
    print('→ Significant difference between content types (p < 0.05)')
else:
    print('→ No significant difference detected')
```

```python
# Spearman correlation: hashtag count vs engagement rate
corr, pval = stats.spearmanr(df['hashtag_count'].dropna(), df.loc[df['hashtag_count'].notna(), 'engagement_rate'])
print(f'Hashtag count ↔ Engagement rate: Spearman r = {corr:.3f}, p = {pval:.4f}')
```

```python
# Caption length vs engagement rate
corr2, pval2 = stats.spearmanr(df['caption_length'].dropna(), df.loc[df['caption_length'].notna(), 'engagement_rate'])
print(f'Caption length ↔ Engagement rate: Spearman r = {corr2:.3f}, p = {pval2:.4f}')
```
