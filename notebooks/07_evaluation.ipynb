# 07 – Model Evaluation & Comparison

Compare Accuracy, Precision, Recall, F1-Score, and Confusion Matrix across trained models.


```python
import sys; sys.path.insert(0, '..')
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

metrics_path = Path('../ml/models/metrics.json')
if metrics_path.exists():
    with open(metrics_path) as f:
        data = json.load(f)
    print('Best Model:', data.get('best'))
    df_metrics = pd.DataFrame({k: v for k, v in data.items() if k != 'best'}).T
    display(df_metrics)
else:
    print('Run ml/train.py first to generate metrics.json')
```
